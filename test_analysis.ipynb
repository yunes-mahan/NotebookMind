```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print('Libraries loaded successfully')
print('NumPy version:', np.__version__)
print('Pandas version:', pd.__version__)
```

```python
# Create a sample sales dataset
np.random.seed(42)

data = {
    'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
              'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'sales': np.random.randint(1000, 5000, size=12),
    'expenses': np.random.randint(500, 3000, size=12),
    'region': ['North', 'South'] * 6
}

df = pd.DataFrame(data)
print(df)
```

```python
# Basic statistics
df['profit'] = df['sales'] - df['expenses']

print('=== Summary Statistics ===')
print(df[['sales', 'expenses', 'profit']].describe())
```

```python
# Group by region and calculate totals
regional_summary = df.groupby('region').agg(
    total_sales=('sales', 'sum'),
    total_expenses=('expenses', 'sum'),
    avg_profit=('profit', 'mean')
).reset_index()

print(regional_summary)
```

```python
# Find the best and worst months
best_month = df.loc[df['profit'].idxmax()]
worst_month = df.loc[df['profit'].idxmin()]

print('Best month: ', best_month['month'], '| Profit:', best_month['profit'])
print('Worst month:', worst_month['month'], '| Profit:', worst_month['profit'])
```

```python
# Plot sales vs expenses over the year
fig, ax = plt.subplots(figsize=(10, 5))

x = range(len(df['month']))
ax.bar(x, df['sales'], width=0.4, label='Sales', color='#7C3AED', alpha=0.8)
ax.bar([i + 0.4 for i in x], df['expenses'], width=0.4, label='Expenses', color='#EF4444', alpha=0.8)

ax.set_xticks([i + 0.2 for i in x])
ax.set_xticklabels(df['month'])
ax.set_title('Monthly Sales vs Expenses', fontsize=14, fontweight='bold')
ax.set_ylabel('Amount ($)')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Calculate a 3-month rolling average of profit
df['rolling_avg_profit'] = df['profit'].rolling(window=3).mean()

print(df[['month', 'profit', 'rolling_avg_profit']].to_string(index=False))
```

```python
# Correlation between sales and expenses
correlation = df['sales'].corr(df['expenses'])
print(f'Correlation between sales and expenses: {correlation:.4f}')

if abs(correlation) > 0.7:
    print('Strong correlation detected')
elif abs(correlation) > 0.4:
    print('Moderate correlation detected')
else:
    print('Weak correlation detected')
```
