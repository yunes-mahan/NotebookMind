```python
# Zelle 3: Funktion
def berechne_fibonacci(n):
    """Gibt die ersten n Fibonacci-Zahlen zurück."""
    fib = [0, 1]
    for i in range(2, n):
        fib.append(fib[-1] + fib[-2])
    return fib[:n]

print(berechne_fibonacci(10))
```

```python
# Zelle 4: Numpy & Datenanalyse
import numpy as np

daten = np.array([4, 7, 13, 2, 9, 11, 5, 8])

print(f'Mittelwert:  {np.mean(daten):.2f}')
print(f'Median:      {np.median(daten):.2f}')
print(f'Std-Abw.:    {np.std(daten):.2f}')
print(f'Max:         {np.max(daten)}')
print(f'Min:         {np.min(daten)}')
```

```python
# Zelle 6: Klassen & OOP
class Student:
    def __init__(self, name, punkte):
        self.name = name
        self.punkte = punkte

    def note(self):
        if self.punkte >= 90: return 'Sehr gut'
        if self.punkte >= 75: return 'Gut'
        if self.punkte >= 60: return 'Befriedigend'
        return 'Nicht bestanden'

    def __repr__(self):
        return f'Student({self.name}, {self.punkte} Pkt → {self.note()})'

studenten = [Student('Max', 92), Student('Lisa', 78), Student('Tom', 55)]
for s in studenten:
    print(s)
```

```python
# Zelle 7: Dictionary & Comprehensions
quadrate = {x: x**2 for x in range(1, 11)}
gerade   = {k: v for k, v in quadrate.items() if k % 2 == 0}

print('Alle Quadrate:', quadrate)
print('Nur gerade:  ', gerade)
```

```python
# Zelle 8: Fehlerbehandlung
def sicheres_teilen(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return 'Fehler: Division durch 0'
    except TypeError as e:
        return f'Typfehler: {e}'

print(sicheres_teilen(10, 2))
print(sicheres_teilen(10, 0))
print(sicheres_teilen(10, 'x'))
```
